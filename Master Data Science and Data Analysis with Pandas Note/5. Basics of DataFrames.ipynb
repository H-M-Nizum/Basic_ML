#### 5.11 Conditional operation on columns
If we want the rows where holiday is greater than 0 in the DataFrame then

```python
df5[df5['holiday'] > 0].head(3)
```

If we want the row with minimum temp

```python
df5[df5['temp'] == df5['temp'].min()]
```

If we want only the name of the temp whose hr is Greter  than 22 then,

```python
df5['temp'][df5['hr'] > 22]
```

if we need two columns like "temp" and "hr"

```python
df5[['temp', 'hr']][df5['hr'] > 22]
```

#### 5.12 accessing row with loc and iloc
Here, loc and iloc are same, this is because we have not set any index here. Index here is by default incremental integers. Loc takes the index value and gives the result for that index, whereas iloc takes the position of an index and gives the row. We can see the difference if we change the default index. In real problems we mostly set the index to the actual data.

`df.loc[<index>]` and `df.iloc[<index position>]`

`df.loc[0]` and `df.iloc[0]`

#### 5.13 Set index
Index by default comes as an incremental integer (0,1…n)

```python
df5.index
```

we can change this to the actual data as:

`df.set_index(“Column Name”)`

```python
df5.set_index('temp').head(2)
```

This is correct but has a problem. This command gives a new DataFrame and
does not change the existing one.

```python
df5.head(2)
```

To make the change effective to the existing one, there are two ways.

1. Store the new DataFrame to the existing one

        `df = df.set_index(“Column name”)`

OR

2. We have a parameter called ‘inplace’ which can be set to true to make the
change effective to the existing DataFrame.
    
        `df.set_index("Column name", inplace=True)`

```python
# df5 = df5.set_index('temp')
df5.set_index('temp', inplace=True)
df5.head(2)
```

Now we can see the difference of loc and iloc also

Now we can see that the loc is taking the index value i.e. temp (.24) where
as iloc gives result for the position of zero (.22).

```python
df5.loc[0.24]
```

```python
df5.iloc[0]
```
