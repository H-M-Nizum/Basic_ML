### 4.1 Creating DataFrame using dictionary data


```python
import pandas as pd
cities = {'Dhaka' : [1,2,3,4], 'Chittagong' : [5,6,7,8]}
df1 = pd.DataFrame(cities)
df1
```

### 4.2 Creating DataFrame using list data

#### 4.2.1 Creating single column
        df = pd.DataFrame(lst)

```python
import pandas as pd
city = ['dhaka', 'chittagong', 'sylhet', 'khulna']
df421 = pd.DataFrame(city)
df421
```

#### 4.2.2 Creating multiple columns
The main point to note here is the length of each list should be same. If
the length does not match, we’ll get an error.

```python
import pandas as pd
name = ['dhaka', 'chittagong', 'sylhet', 'khulna']
area = [1200, 2000, 3000, 4000]
df422 = pd.DataFrame(list(zip(name, area)))
df422
```

#### 4.3 Adding custom column name
This is done by a parameter ‘column’. 

df = pd.DataFrame(data, columns=[‘col1, ‘col2])

```python
df43 = pd.DataFrame(list(zip(name, area)), columns = ['Name', 'Area'])
df43
```

#### 4.4 creating DataFrame from list of dictionaries
Each element (dict) of the list will become a row and the dictionary key value pair will become
column and its value.

```python
list_dict = [{'Name': 'dhaka', 'Area': 1200}, {'Name': 'chittagong', 'Area': 2000}, {'Name': 'sylhet', 'Area': 3000}, {'Name': 'khulna', 'Area': 4000}]
df44 = pd.DataFrame(list_dict)
df44
```
